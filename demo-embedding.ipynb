```python
from ollama import embeddings
from numpy import array, dot
from numpy.linalg import norm


MODEL = "qwen3-embedding:0.6b"


def cosine_similartiy(a, b):
    return round( dot(a, b) / (norm(a) * norm(b)), 2)


prompt = "Schreibe einen Text über Herbst und Blätter."
text_2 = "Im Herbst fallen immer wieder Blätter zu Boden."
text_3 = "Beim Rennen quietschten die Reifen der Autos."
```

```python
response_p = embeddings(model=MODEL, prompt=prompt)
response_2 = embeddings(model=MODEL, prompt=text_2)
response_3 = embeddings(model=MODEL, prompt=text_3)

emb_p = response_p.embedding
emb_2 = response_2.embedding
emb_3 = response_3.embedding
```

```python
print(
    "Wie ähnlich sind Prompt und Dokument 2?",
    cosine_similartiy(emb_p, emb_2)
)

print(
    "Wie ähnlich sind Prompt und Dokument 3?",
    cosine_similartiy(emb_p, emb_3)
)
```
