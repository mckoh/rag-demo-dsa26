```python
from ollama import chat


MODEL = "llama3.1:latest"

messages = [
    {"role": "user", "content": "Schreibe ein Herbstgedicht mit 5 Absätzen."}
]
```

```python
response = chat(model=MODEL, messages=messages)
```

```python
print(response.message.content)
```

## Kontext weiter schreiben

```python
# Response des Systems in den Messages ergänzen
messages.append(
    {"role": "assistant", "content": response.message.content}
)
```

```python
# ev. nächste User-Message ergänen
messages.append(
    {"role": "user", "content": "Kürze das Gedicht auf 1 Absatz."}
)
```

```python
# Neuen, erweiterten Kontext ins Modell schicken und weiter generieren lassen
chat(
    model=MODEL,
    messages=messages
)
```
